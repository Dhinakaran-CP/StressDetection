# Phase 1: Raw Data Synchronization Audit
This notebook scans the raw dataset directories, identifies modality presence, and checks duration synchronization across Video, Audio, and Physiological signals.

```python
print("[BACKGROUND] Initializing system environment and path imports...")
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath('../'))
from src.data.loader import RawDataLoader
print("[STATUS] System paths configured successfully.")
```

```python
print("[BACKGROUND] Checking directory configuration...")
base_dir = '/content/drive/MyDrive/Multimodal_stress_Detection/' if os.path.exists('/content/drive') else '../../'
print(f"[INFO] Target base directory detected: {os.path.abspath(base_dir)}")

print("[BACKGROUND] Instantiating RawDataLoader and scanning folders recursively...")
loader = RawDataLoader(base_dir)
summary, detailed_df = loader.run_synchronization_audit()
print("[STATUS] Scanning complete. Multimodal keys grouped successfully.")
```

```python
print("[BACKGROUND] Formatting raw alignment metrics summary...")
print("\n==================================================")
print("RAW DATA SYNCHRONIZATION AUDIT REPORT")
print("==================================================")
for k, v in summary.items():
    print(f"{k:25}: {v}")
print("==================================================")
```

```python
print("[BACKGROUND] Preparing directory paths for CSV audit reports...")
os.makedirs('../reports/synchronization/', exist_ok=True)

print("[BACKGROUND] Writing file-level audit details to raw_files_alignment_audit.csv...")
detailed_df.to_csv('../reports/synchronization/raw_files_alignment_audit.csv', index=False)

print("[BACKGROUND] Writing metric summary to synchronization_report.csv...")
report_df = pd.DataFrame([{"metric": k, "value": v} for k, v in summary.items()])
report_df.to_csv('../reports/synchronization/synchronization_report.csv', index=False)

print("[STATUS] Reports successfully saved under early_fusion/reports/synchronization/")
```
