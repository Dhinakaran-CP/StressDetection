# Phase 8: Final Model Selection & Comparison
Summarizes parameter sizes, model inference results, and final synchronization metrics.

```python
print("[BACKGROUND] Collecting final comparison data...")
import os
import pandas as pd

eval_csv = '../reports/evaluation/robustness_metrics.csv'
if os.path.exists(eval_csv):
    df = pd.read_csv(eval_csv)
    # Sort for cleaner, structured presentation by model and scenario
    df_sorted = df.sort_values(by=['Model', 'Scenario'])
    
    # Print to console
    print("\n=============================================================================================================================")
    print("STAGE 8: MULTIMODAL ROBUSTNESS & ABLATION COMPARISON STUDY (ALL CLASSES & MODALITIES)")
    print("=============================================================================================================================")
    print(df_sorted.to_string(index=False))
    print("=============================================================================================================================")
    
    # Export beautiful markdown report
    md_table = df_sorted.to_markdown(index=False)
    report_md = f"""# Stage 8: Multimodal Ablation Study & Robustness Comparison\n\n"
    report_md += "This report summarizes the comparative performance, class-specific metrics, architectural sizes, and inference latency costs of the five trained architectures under various sensor dropout (missing modality) scenarios.\n\n"
    report_md += f"\n{md_table}\n"
    os.makedirs('../reports/evaluation/', exist_ok=True)
    with open('../reports/evaluation/ablation_study_report.md', 'w') as f:
        f.write(report_md)
    print("[STATUS] Formatted ablation report successfully saved to reports/evaluation/ablation_study_report.md")
else:
    print("[WARNING] Robustness CSV report not found. Execute Notebook 04 first.")

print("[STATUS] Project final select reporting finished successfully.")
```
